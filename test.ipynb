````python
TOMATO_TEST = [
    [["PLANT", "MELON"], ["WATER"]],
    [],
    [["WATER"]],
    [],
    [["WATER"]],
    [],
    [["WATER"]],
    [["WATER"]],
    [["WATER"]],
    [["WATER"]],
    [["WATER"], ["HARVEST"]],
]


def do_nothing(obs):
    step = obs.get("step", 0)
    day, step_in_day = step // 24, step % 24
    if step_in_day == 0:
        return {"farmer": ["PASS"], "hands": [], "market": [["BUY_SEED", "MELON", 1], ["BUY_PRODUCT", "FERTILIZER", 1]]}
    if day < len(TOMATO_TEST) and step_in_day <= len(TOMATO_TEST[day]):
        farmer_act = TOMATO_TEST[day][step_in_day - 1]
        print(f"Day {day}, Step {step_in_day}: Farmer action: {farmer_act}")
        return {
            "farmer": farmer_act,
            "hands": [],
            "market": []
        }
    return {"farmer": ["WEST"], "hands": [], "market": []}

def test(obs):
    step = obs.get("step", 0)
    day, step_in_day = step // 24, step % 24

    # Kiểm tra nếu ngày nằm trong phạm vi dữ liệu
    if day < len(FARMER_ACTIONS):
        farmer_act = FARMER_ACTIONS[day][step_in_day] if step_in_day < len(FARMER_ACTIONS[day]) else ["PASS"]
        market_act = MARKET_ACTIONS[day][step_in_day] if step_in_day < len(MARKET_ACTIONS[day]) else []

        # Lấy hành động của từng hand tại step_in_day
        hands_act = []
        if day < len(HANDS_ACTIONS):
            for hand in HANDS_ACTIONS[day]:
                if step_in_day < len(hand):
                    hands_act.append(hand[step_in_day])
                else:
                    hands_act.append(["PASS"])

        return {
            "farmer": farmer_act,
            "hands": hands_act,
            "market": market_act
        }
        
    return {"farmer": ["PASS"], "hands": [], "market": []}
````

````python
from kaggle_environments import make

config_non_weed = {
    "weedSpawnChance": 0.0,
}

env = make("kaggriculture", config_non_weed, debug=True)
env.run([do_nothing, test])

final = env.steps[-1]
for i, s in enumerate(final):
    print(f"Player {i}: reward={s.reward}, status={s.status}")

# Lấy chuỗi HTML thay vì để nó tự hiển thị (return_obj=True hoặc dùng mode="html")
html_content = env.render(mode="html", width=800, height=600)

with open("kaggriculture_replay.html", "w", encoding="utf-8") as f:
    f.write(html_content)
````
